```python
a=int(input("Enter a value:"))

if a>0:
  print("It's a possitive number")
```

```python
ch=input("Enter a character:")

if ch in vowel:
  print("It's a vowel")
else:
  print("It is a consonant")
```

```python
a=int(input("Enter a value:"))

if a>0:
  print("It's a possitive number")
elif a<0:
  print("It is a neg number")
else:
  print("It is a zero")
```
