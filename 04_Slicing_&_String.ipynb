***Slice***

datavariable[start:stop:step=1]

by default start =0 and step =1

slicing can be applied in list,string,tuple,set



```python
lst=[11,12,13,14,15,16,17,18,19]
print(lst[0:4:1])   #2 types indexing forward ,backward
```

```python
print(lst[-1:4:-1])
print(lst[-1:4:1])
print(lst[3:6:-1])
```

```python
print(lst[-1:4:1])
```

```python
print(lst[::])
```

```python
print(lst[::-1])
```

```python
print(lst[:3:-1])
```

```python
print(lst[:-6:-2])
```

```python
rev=lst[::-1]
if lst ==rev:
  print("pallindrom")
else:
  print("Not pallindrom")
```

```python
string=input("Enter string:")
rev_str=string[::-1]
if string ==rev_str:
  print("pallindrome")
else:
  print("Not pallindrome")
```

The password must contain at least 8 characters.

It must contain at least 2 uppercase letters.

It must contain at least 2 digits.

It must contain at least 1 special character.

It must not contain any whitespace.

It must contain at most 6 lowercase letters.

```python

pasw=input("Enter password:")
ln=len(pasw)
up,lw,dg,sp,sc=0,0,0,0,0

for ch in pasw:
  if ch.isupper():
    up+=1
  elif ch.islower():
    lw+=1
  elif  ch.isdigit():
    dg+=1
  elif  ch.isspace():
    sp+=1
  elif not(ch.isspace() or ch.isalnum()):
    sc+=1
if ln>=8 and up>=2 and lw<=6 and dg>=2 and sp==0 and sc>=1:
  print("Strong password")
else:
  print("Weak password")
```

```python
s=input("Enter string:")
vowel="aeiouAEIOU"
up,lw,dg,vw,cns="","","","",""
for ch in s:
  if ch in vowel:
   vw+=ch
  if ch not in vowel and ch.isalpha():
    cns+=ch
  if ch.isupper():
    up+=ch
  elif ch.islower():
    lw+=ch
  elif ch.isdigit():
    dg+=ch


print(up  ,lw ,dg, vw, cns)
```
