Adding element inro a list

1.append():add one element at the end of the list

2.extend():add two list

3.insert():insert a value at a specific index

```python
lst=[1,2,3,4,5]
print(lst)
lst.append(7)
print(lst)
lst.extend([6,7])
print(lst)
lst.insert(2,9)
print(lst)
lst.extend([])
```

Removing Elements

1.remove():removes a specific value

2.pop(): pop out a value from an index

3.clear(): clear a whole list

```python
lst=[11,12,13,14,15,16]
print(lst)
lst.remove((12))
print(lst)
temp=lst.pop(2)
print(temp)
print(lst)

lst.clear
print(lst)
```
